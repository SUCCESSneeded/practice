**3D Surface Plot**

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)

xx,yy=np.meshgrid(x,y)
#yy.shape
z=np.sin(xx)+np.cos(yy)
z.shape

fig=plt.figure(figsize=(10,10))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='jet')
```

**Contour**

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)

xx,yy=np.meshgrid(x,y)
#yy.shape
z=xx**2+yy**2
z.shape

fig=plt.figure(figsize=(10,10))
ax=plt.subplot(2,1,1)
ax.contour(xx,yy,z)
ax1=plt.subplot(2,1,2)
ax1.contourf(xx,yy,z)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Creating the match data dictionary
data = {
    "matches": list(range(1, 21)),  # Match numbers 1 to 20
    "runsovers": [
        45, 12, 88, 4, 102, 0, 35, 67, 18, 54,
        71, 9, 115, 23, 44, 61, 8, 82, 39, 50
    ],  # Runs scored in each match
    "balls": [
        30, 15, 55, 3, 62, 1, 28, 40, 22, 38,
        50, 12, 75, 18, 35, 42, 10, 58, 31, 36
    ],  # Balls faced in each match
    "fours": [
        5, 1, 9, 0, 12, 0, 4, 7, 2, 6,
        8, 1, 14, 3, 4, 5, 1, 9, 3, 5
    ],  # Fours hit per match
    "sixes": [
        1, 0, 3, 0, 4, 0, 0, 2, 0, 1,
        2, 0, 5, 0, 1, 2, 0, 3, 1, 1
    ],  # Sixes hit per match
    "strike_rate": [
        150.0, 80.0, 160.0, 133.3, 164.5, 0.0, 125.0, 167.5, 81.8, 142.1,
        142.0, 75.0, 153.3, 127.8, 125.7, 145.2, 80.0, 141.4, 125.8, 138.9
    ]  # Strike rate for each match
}

df=pd.DataFrame(data)

tempdf=df[(df['fours'].isin([0,1,2,3,4,5])) & (df['sixes'].isin([0,1,2,3,4,5]))]
grid=tempdf.pivot_table(values=['strike_rate'],index='matches',columns=['fours','sixes'],)
plt.figure(figsize=(10,5))
plt.imshow(grid)
plt.colorbar()
```
