```python
# all subarrays of the list
arr=[1,2,3]
list1=[]
for i in range(len(arr)):
  list2=[]
  for j in range(i,len(arr)):
    list2.append(arr[j])
    list1.append(list2.copy())
print(list1)
```

```python
#subsets
nums=[1,2,3]
ans=[]
def backtrack(i,current):
     if i == len(nums):
       ans.append(current.copy())
       return
     backtrack(i+1,current)
     current.append(nums[i])
     backtrack(i+1,current)
     current.pop()
backtrack(0,[])
print(ans)
```
